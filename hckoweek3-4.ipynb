```python
import numpy as np

np.random.seed(0)
X = np.random.rand(100, 2)
y = 3 * X[:, 0] + 4 * X[:, 1] + 1 + np.random.randn(100) * 0.1

X_b = np.c_[np.ones((100, 1)), X]
theta = np.random.randn(3, 1)

def compute_mse(X, y, theta):
    m = len(y)
    predictions = X.dot(theta)
    errors = predictions - y
    return (1 / (2 * m)) * (errors.T.dot(errors))

def compute_gradient(X, y, theta):
    m = len(y)
    predictions = X.dot(theta)
    errors = predictions - y
    return (1 / m) * (X.T.dot(errors))

def compute_hessian(X):
    m = len(X)
    return (2 / m) * (X.T.dot(X))

def get_eigenvalues(X):
    hessian = compute_hessian(X)
    eigenvals, _ = np.linalg.eig(hessian)
    return eigenvals

theta_history = []
learning_rate = 0.1
iterations = 1000

for i in range(iterations):
    grad = compute_gradient(X_b, y, theta)
    theta = theta - learning_rate * grad
    theta_history.append(theta.copy())

final_mse = compute_mse(X_b, y, theta)
eigen_vals = get_eigenvalues(X_b)

print(f"Final Theta: {theta.flatten()}")
print(f"Final MSE: {final_mse}")
print(f"Eigenvalues of Hessian: {eigen_vals}")
```
